```python
text = """Natural Language Processing (NLP) is a subfield of linguistics, computer science, and artificial
intelligence concerned with the interactions between computers and human language. It's used to
analyze text, allowing machines to understand, interpret, and manipulate human language. NLP has
many real-world applications, including machine translation, sentiment analysis, and chatbots."""
```

```python
# 1. Tokenization: Split the text into individual words (tokens).
token = text.split()
print(token)
```

```python
# 2. Lowercasing: Convert all tokens to lowercase.
lowercase = []
for word in token:
  lowercase.append(word.lower())
print(lowercase)
```

```python
# 3. Punctuation Removal: Remove all punctuation marks from the tokens.
import string

punctuation_removal = []
for word in lowercase:
  clean_word = ""
  for w in word:
    if w not in string.punctuation:
      clean_word += w
  if clean_word != "":
    punctuation_removal.append(clean_word)
print(punctuation_removal)
```

```python
# 4. Stop Word Removal: Remove common stop words (e.g., "the", "is", "and").
stop_words = ["the", "a", "an", "in", "on", "at", "for", "to", "of", "and", "is", "are"]
removal = []
for w in punctuation_removal:
  if w not in stop_words:
    removal.append(w)
print(removal)
```

```python
# 5. Stemming: Reduce words to their root form using a simple algorithm.
stemming = []
for words in removal:
  if words.endswith("ing") and len(words) > 4:
    stemming.append(words[:-3])
  elif words.endswith("ed") and len(words) > 3:
    stemming.append(words[:-2])
  elif words.endswith("s") and len(words) > 3:
    stemming.append(words[:-1])
  else:
    stemming.append(words)
print(stemming)
```

```python
# Bonus. Implement a simple lemmatization function that uses a dictionary to map common words to their base form (e.g., "is" -> "be", "are" -> "be"). Compare the results with your stemming function.
lemmatization = {"is" : "be",
                 "are" : "be"}
l = []
for char in punctuation_removal:
  if char in lemmatization:
    l.append(lemmatization[char])
  else:
    l.append(char)
print(l)

```
