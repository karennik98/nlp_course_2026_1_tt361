```python
import string


raw_text = """Natural Language Processing (NLP) is a subfield of linguistics, computer science, and artificial
intelligence concerned with the interactions between computers and human language. It's used to
analyze text, allowing machines to understand, interpret, and manipulate human language. NLP has
many real-world applications, including machine translation, sentiment analysis, and chatbots."""


tokens_step1 = raw_text.split()


tokens_step2_3 = []
for token in tokens_step1:

    clean_token = "".join([char.lower() for char in token if char not in string.punctuation])
    if clean_token:
        tokens_step2_3.append(clean_token)


stop_words = ["the", "a", "an", "in", "on", "at", "for", "to", "of", "and", "is", "are"]
tokens_step4 = [token for token in tokens_step2_3 if token not in stop_words]


def simple_stemmer(word):
    """
    A simple rule-based stemmer removing common English suffixes.
    """
    suffixes = ['ing', 'ion', 'ed', 'ly', 'es', 's']
    for suffix in suffixes:


        if word.endswith(suffix) and len(word) - len(suffix) >= 3:
            return word[:-len(suffix)]
    return word

tokens_step5 = [simple_stemmer(token) for token in tokens_step4]


lemma_dict = {
    "is": "be",
    "are": "be",
    "has": "have",
    "its": "it",
    "used": "use",
    "allowing": "allow",
    "interactions": "interaction",
    "computers": "computer",
    "machines": "machine",
    "applications": "application",
    "chatbots": "chatbot",
    "linguistics": "linguistics"
}

def simple_lemmatize(word):
    """
    Lemmatize words using a predefined dictionary mapping.
    """
    return lemma_dict.get(word, word)

tokens_bonus_lemmatized = [simple_lemmatize(token) for token in tokens_step4]


print("=== 1. Tokens after Step 1 (Tokenization) ===")
print(tokens_step1)
print(f"\nTotal tokens: {len(tokens_step1)}\n")

print("=== 2. Tokens after Steps 2 & 3 (Lowercasing & Punctuation Removal) ===")
print(tokens_step2_3)
print(f"\nTotal tokens: {len(tokens_step2_3)}\n")

print("=== 3. Tokens after Step 4 (Stop Word Removal) ===")
print(tokens_step4)
print(f"\nTotal tokens: {len(tokens_step4)}\n")

print("=== 4. Tokens after Step 5 (Simple Stemming) ===")
print(tokens_step5)
print(f"\nTotal tokens: {len(tokens_step5)}\n")

print("=== Bonus: Tokens after Lemmatization ===")
print(tokens_bonus_lemmatized)
```
