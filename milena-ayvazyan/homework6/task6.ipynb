```python
import re
import sys
import random
```

```python
try:
  all_lines = []
  with open("robert_frost (1).txt", "r", encoding="utf-8") as f:
    for line in f:
      clean = line.lower()
      clean = re.sub(r'[^\w\s]', '', clean)
      words = clean.split()
      if not words:
        continue
      words.append('END')
      all_lines.append(words)


except OSError as e:
  print(f'Error reading file: {e}')
  sys.exit(1)
```

```python
start_counts = {}
for i in all_lines:
  first_word = i[0]
  if first_word not in start_counts:
    start_counts[first_word] = 1
  else: start_counts[first_word] += 1
```

```python
start_probs = {}
for key, value in start_counts.items():
  start_probs[key] = value / len(all_lines)
```

```python
first_order = {}
for line_words in all_lines:
  for i in range(len(line_words) - 1):
    prev_word = line_words[i]
    next_word = line_words[i+1]
    if prev_word not in first_order:
      first_order[prev_word] = {}
    if next_word not in first_order[prev_word]:
       first_order[prev_word][next_word] = 1
    else: first_order[prev_word][next_word] += 1
```

```python
first_probs = {}
for key, value in first_order.items():
  all_sum = sum(value.values())
  first_probs[key] = {}
  for second_word, second_value in value.items():
    first_probs[key][second_word] = second_value / all_sum
```

```python
second_order = {}
for line_words in all_lines:
  for i in range(len(line_words) - 2):
    context = (line_words[i], line_words[i + 1])
    next_word = line_words[i+2]
    if context not in second_order:
      second_order[context] = {}
    if next_word not in second_order[context]:
      second_order[context][next_word] = 1
    else: second_order[context][next_word] += 1
```

```python
second_probs = {}
for key, value in second_order.items():
  all_s = sum(value.values())
  second_probs[key] = {}
  for third_word, third_value in value.items():
    second_probs[key][third_word] = third_value / all_s
```

```python
def deterministic_lines(start_probs, first_probs,second_probs, num_lines = 4):
  poem_lines = []

  for _ in range(num_lines):
    first_word = max(start_probs, key=start_probs.get)
    line = [first_word]

    while len(line) < 20:

      if len(line) == 1:
        key = line[-1]
        table = first_probs
      else:
        key = (line[-2], line[-1])
        table = second_probs
      if key not in table:
        break

      next_word = max(table[key], key=table[key].get)
      if next_word == "END":
        break
      line.append(next_word)

    poem_lines.append(" ".join(line))
  return "\n".join(poem_lines)

print(deterministic_lines(start_probs, first_probs,second_probs, num_lines = 4))
```

```python
def pick_cumulative(distribution):
  r = random.random()
  cumulative = 0
  for k, v in distribution.items():
    cumulative += v
    if r <= cumulative:
      return k
  return k

def cumulative_lines(start_probs, first_probs,second_probs, num_lines = 4):
  poem_lines = []

  for _ in range(num_lines):
    first_word = pick_cumulative(start_probs)
    line = [first_word]

    while len(line) < 20:

      if len(line) == 1:
        key = line[-1]
        table = first_probs
      else:
        key = (line[-2], line[-1])
        table = second_probs
      if key not in table:
        break

      next_word = pick_cumulative(table[key])
      if next_word == "END":
        break
      line.append(next_word)

    poem_lines.append(" ".join(line))
  return "\n".join(poem_lines)

print(cumulative_lines(start_probs, first_probs,second_probs, num_lines = 4))
```
