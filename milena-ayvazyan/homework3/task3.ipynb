```python
import pandas as pd
import re
import nltk
```

```python
# Download necessary NLTK corpora and tokenizer models for text processing
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
# Load the Coachella tweets dataset with proper encoding
df = pd.read_csv("Coachella-2015-2-DFE.csv", encoding='latin-1')
df
```

```python
df.info()
```

**Find** all the hashtags in the tweets, such as #coachella, #beychella, etc. Extract them and save them in a separate column named "hashtags".

```python
df['hashtags'] = df['text'].apply(lambda x: re.findall(r'#\w+', x))
df['hashtags']
```

Find all the emails in the tweets, such as example@gmail.com, user@yahoo.com, etc. Extract them and save them in a separate column named "emails".**bold text**

```python
df['emails'] = df['text'].apply(lambda x: re.findall(r'[\w.]+@\w+\.[A-Za-z]{2,}', x))
df['emails']
```

**Write the following functions to remove unwanted elements from the tweets and apply them to the new tweets text column:**

**remove_usernames:** *This function should remove any usernames that start with @, such as @coachella, @beyonce, etc.*

```python
def remove_usernames(text):
  return re.sub(r'@\w+', '', str(text))
```

**remove_links:** *This function should remove any links that start with http or https, such as https://www.coachella.com/, http://bit.ly/2GdOZLz, etc.*

```python
def remove_links(text):
  return re.sub(r'\b(?:https|http)\S+', '', str(text))
```

**remove_non_ascii_symbols:** *This function should remove any symbols that are not part of the ASCII character set, such as emojis, accented letters, etc.*

```python
def remove_non_ascii_symbols(text):
  return re.sub(r'[^\x00-\x7F]', '', text)
```

**to_lower:** *This function should convert all the letters in the tweet text to lower case, such as COACHELLA to coachella, Beyoncé to beyoncé, etc.*

```python
def to_lower(text):
  return text.lower()
```

**remove_stop_words:** *This function should remove any common words that do not add much meaning to the tweet text, such as the, a, an, and, etc.*

```python
def remove_stop_words(text):
  token = nltk.word_tokenize(text)
  stopwords = nltk.corpus.stopwords.words("english")
  result = [t for t in token if t.lower() not in stopwords]
  return " ".join(result)
```

**remove_digits:** *This function should remove any numbers in the tweet text, such as 2018, 10, 100, etc.*

```python
def remove_digits(text):
  return re.sub(r'\d+', '', text)
```

**remove_special_characters:** This function should remove any punctuation marks or other special characters in the tweet text, such as . , ! ? # $ % & * ( ) + = - _ [ ] { } ; : ' " / \ | < > ` ~, etc. *italicized text*

```python
def remove_special_characters(text):
  return re.sub(r'[^a-zA-Z0-9\s]', '', text)
```

```python
df["cleaned_text"] = df["text"]

df["cleaned_text"] = df["cleaned_text"].apply(remove_usernames)
df["cleaned_text"] = df["cleaned_text"].apply(remove_links)
df["cleaned_text"] = df["cleaned_text"].apply(remove_non_ascii_symbols)
df["cleaned_text"] = df["cleaned_text"].apply(to_lower)
df["cleaned_text"] = df["cleaned_text"].apply(remove_stop_words)
df["cleaned_text"] = df["cleaned_text"].apply(remove_digits)
df["cleaned_text"] = df["cleaned_text"].apply(remove_special_characters)
df["cleaned_text"]
```
