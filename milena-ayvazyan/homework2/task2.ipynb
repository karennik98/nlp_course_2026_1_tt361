```python
import re
```

Problem 1: Extract Email Addresses

```python
text = "Contact us at support@example.com or sales@company.org for assistance.For personal inquiries, email john.doe123@university.edu."
matches = re.findall(r'[\w.]+@\w+\.com|[\w.]+@\w+\.org|[\w.]+@\w+\.edu', text)
matches
```

Problem 2: Validate Phone Numbers

```python
phone_numbers = "Valid: 123-456-7890, 987-654-3210 Invalid: 12-345-67890, 1234567890, 123-45-6789"
matches = re.findall(r'\d{3}-\d{3}-\d{4}', phone_numbers)
print(f'Valid phone numbers: {matches}')
```

Problem 3: Extract Dates

```python
dates = "Important dates: 25/12/2023, 01-01-2024, 31/05/2023, and 15-10-2024."
matches = re.findall(r'\d{2}[/|-]\d{2}[/|-][1-9]\d{3}', dates)
matches
```

Problem 4: Find Repeated Words

```python
repeated_words = "The the quick brown fox jumps over the the lazy dog."
matches = re.findall(r'\b((\w+)\s+\2)\b', repeated_words, re.IGNORECASE)
print([i[0] for i in matches])
```

Problem 5: Extract Hashtags

```python
text = "Check out our new products: #Sale2024, #NewArrival, and #Discounts!"
matches = re.findall(r'#\w+', text)
print(matches)
```

Problem 6: Validate Password Strength

```python
text = "Valid: Password123, Secure456 Invalid: weak, password, Password"
matches = re.findall(r'(?=.*[A-Z]{1,9})(?=.*[a-z]{1,9})(?=.*\d+)\w{8,}', text)
print(matches)
```

Problem 7: Extract URLs

```python
text = "Visit our website at https://www.example.com or check out http://blog.example.org for updates."
matches = re.findall(r'\b(?:https|http)\S+', text)
matches
```

Problem 8: Replace Multiple Spaces with a Single Space

```python
text = "This   text    has   multiple    spaces."
matches = re.sub(r'\s+', ' ', text)
matches
```

Problem 9: Extract Quoted Text

```python
text = 'He said, "Hello, world!" and she replied, "Hi there!"'
matches = re.findall(r'"([^"]+)"', text)
for m in matches:
  print(m)
```

Problem 10: Validate IP Addresses

```python
text = "Valid: 192.168.1.1, 10.0.0.255 Invalid: 256.1.2.3, 192.168.01.1, 192.168.1"
matches = re.findall(r'\b(?:(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]\d|\d)\.){3}(?:25[0-5]|2[0-4]\d|1\d{2}|[1-9]\d|\d)\b', text)
print("Valid IP addresses:")
for m in matches:
  print(m)
```
