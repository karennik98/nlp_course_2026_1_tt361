```python
import pandas as pd
import nltk
import re
import string
```

```python
df = pd.read_csv("Coachella-2015-2-DFE.csv",encoding='windows-1256')
```

```python
def extract_hashtags(text):
    return re.findall(r'#\w+',text)

def extract_email(text):
    return re .findall(r"[\w./*-=+#$]+@[\w./*-=+#$]+\.[a-zA-Z]{2,3}",text)
    
```

```python
df["hashtags"] = df["text"].apply(extract_hashtags)
print(df["hashtags"])
```

```python
df["emails"] = df["text"].apply(extract_email)
```

```python
nltk.download("stopwords")
nltk.download("punkt")
nltk.download("punkt_tab")
stop_words = nltk.corpus.stopwords.words("english")
```

```python
def remove_username(text):
    return re.sub(r"@\w+","",text)

def remove_url(text):
    return re.sub(r"https?://\S+", "", text)

def remove_non_ascii(text):
    return text.encode("ascii", "ignore").decode("ascii")

def remove_stopwords(tweets):
    tokens = nltk.word_tokenize(tweets)
    filtered_tokens = [
        token for token in tokens
        if token.lower() not in stop_words
    ]

    return " ".join(filtered_tokens)

def remove_digites(text):
    return re.sub(r"\d+","",text)

def remove_punctuation(text):
    tokens = nltk.word_tokenize(text)
    
    tokens = [word for word in tokens if word not in string.punctuation]
    
    return " ".join(tokens)    
```

```python
df["tweets"] = df["text"].apply(remove_username)
df["tweets"] = df["tweets"].apply(remove_url)
df["tweets"] = df["tweets"].apply(remove_non_ascii)
df["tweets"] = df["tweets"].str.lower()
df["tweets"] = df["tweets"].apply(remove_stopwords)
df["tweets"] = df["tweets"].apply(remove_digites)
df["tweets"] = df["tweets"].apply(remove_punctuation)


print(df["tweets"])
```
