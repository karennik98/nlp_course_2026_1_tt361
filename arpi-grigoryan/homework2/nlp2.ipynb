```python
import re
```

```python
#1
class ExtractEmail:
    def __init__(self,text):
        self.text = text
    def extract(self):
        pattern = r"[\w./*-=+#$]+@[\w./*-=+#$]+\.[a-zA-Z]{2,3}"  
        match = re.findall(pattern,self.text)
        return match

text = input("Enter the text")
extractor = ExtractEmail(text)
print(extractor.extract())

    
```

```python
#2
class ValidNumbers:
    def __init__(self,number):
        self.number = number
    def find(self):
        pattern = r"\d{3}-\d{3}-\d{4}"
        match = re.findall(pattern,self.number)
        return match
    
number = input()
valid = ValidNumbers(number)
print(valid.find())    
```

```python
#3
class ExtractDates:
    def __init__(self,date):
        self.date = date
    def extract(self):
        pattern = r"\d{2}-\d{2}-\d{4}|\d{2}\.\d{2}\.\d{4}"
        m = re.findall(pattern,self.date)
        return m
    
date = input()
extractor = ExtractDates(date)
print(extractor.extract())    
```

```python
#4
class FindRepeatedWords:
    def __init__(self, word):
        self.word = word

    def FindWord(self):
        pattern = r"\b(\w+)\s+(\1)\b"
        m = re.findall(pattern, self.word)
        result = []

        for word1, word2 in m:
            result.append(word1)
            result.append(word2)

        return result

text = input("Enter the text: ")
finder = FindRepeatedWords(text)
result = finder.FindWord()
print("Repeated words:", *result)
       
```

```python
#5
class ExtractHashtag:
    def __init__(self,hash):
        self.hash = hash
    def extract(self):
        pattern = r"#\w+"
        m = re.findall(pattern,self.hash)
        return m
    
text = input()
extractor = ExtractHashtag(text)
print(extractor.extract())  
```

```python
class PasswordValidator:
    def __init__(self, passwords):
        self.passwords = passwords

    def validate(self):
        pattern = r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d).{8,}$"

        valid_passwords = []

        for password in self.passwords:
            if re.match(pattern, password):
                valid_passwords.append(password)

        return valid_passwords


passwords = ["Password123", "Secure456", "weak", "password", "Password"]
validator = PasswordValidator(passwords)
result = validator.validate()
print("Valid passwords:", *result)
```

```python
#7
class GetURL:
    def __init__(self,url):
        self.url = url
    def extract(self):
        pattern = r"https?://\S+"
        m = re.findall(pattern,self.url)
        return m
    
text = input()
extractor = GetURL(text)
print(extractor.extract()) 
```

```python
#8
class GetSpace:
    def __init__(self,space):
        self.space = space
    def extract(self):
        pattern = r"\s+"
        m = re.sub(pattern," ",self.space)
        return m
    
text = input()
extractor = GetSpace(text)
print(extractor.extract()) 
```

```python
#9
class ExtractQuotedText:
    def __init__(self, text):
        self.text = text

    def extract(self):
        pattern = r'"([^"]*)"'
        m = re.findall(pattern, self.text)
        return m


text = input("Enter the text: ")
extractor = ExtractQuotedText(text)
result = extractor.extract()
print("Quoted text:", *result)
```

```python
import re
class IPValidator:
    def __init__(self, text):
        self.text = text

    def validate(self):
        pattern = r"\b(?:(?:25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)\.){3}(?:25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)\b"
        ips = re.findall(pattern, self.text)
        return ips


text = input("Enter the text: ")
validator = IPValidator(text)
result = validator.validate()
print("Valid IP addresses:", *result)

```
